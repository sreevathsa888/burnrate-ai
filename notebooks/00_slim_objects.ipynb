# 00 — Slim objects.csv

`objects.csv` is huge (278 MB, ~462k rows, includes long free-text columns like
`description`/`overview`). We only need 12 columns for the survival model, so
we select them **at read time** with `usecols` — this avoids ever loading the
big text columns into memory, which is much faster than loading everything
and dropping columns after.

Run this once. Output: `data/raw/objects_slim.csv` (should land around 15-25 MB).

```
import pandas as pd

RAW = "../data/raw"

cols = ["id", "entity_type", "name", "category_code", "status", "founded_at",
        "closed_at", "country_code", "first_funding_at", "last_funding_at",
        "funding_rounds", "funding_total_usd"]

obj = pd.read_csv(f"{RAW}/objects.csv", encoding="ISO-8859-1",
                   usecols=cols, low_memory=False)

print(f"[read] {len(obj):,} total rows, {obj['entity_type'].nunique()} entity types")
print(obj["entity_type"].value_counts())
```

```
companies = obj[obj["entity_type"] == "Company"].copy()
print(f"[filter] {len(companies):,} companies kept")

companies.to_csv(f"{RAW}/objects_slim.csv", index=False)
print(f"[done] wrote {RAW}/objects_slim.csv")
```
